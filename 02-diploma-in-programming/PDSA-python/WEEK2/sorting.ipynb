### Naive Search by scanning the list

```python
def naivesearch(v,l):
    for i in l:
        if i == v:
            return(True)
    return(False)
```

### Binary Search

```python
def binarysearch(v,l):
    if l == []:
        return(False)

    m = len(l)//2

    if v == l[m]:
        return(True)

    if v < l(m):
        return(binarysearch(v,l[:m]))
    else:
        return(binarysearch(v,l[m+1:]))
```
